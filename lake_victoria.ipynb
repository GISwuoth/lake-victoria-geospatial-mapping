```python
from pathlib import Path

PROJECT_DIR = Path.cwd()
DATA_FILE = PROJECT_DIR/"data"/"lake_victoria.geojson"

lake = gpd.read_file(DATA_FILE)

print("CRS: ", lake.crs)
print("Rows & Columns: ", lake.shape)
print("Geometry types: ", lake.geom_type.unique())
print("Bounds: ", lake.total_bounds)
```

```python
# Reprojection for applying buffer
#read.to_crs("EPSG:32643", inplace = True)
```

```python
m = leafmap_folium.Map(
    center = [-1.0, 33.0],
    zoom = 7,
    basemap = "OpenStreetMap"
)
m.add_gdf(
    lake,
    layer_name = "Lake Victoria Boundary",
    style = {
        "color": "red",
        "fillColor": "skyblue",
        "weight": 2,
        "fillOpacity": 0.35
    }
)


m 
```

```python
from pathlib import Path

output_dir = Path.cwd()/"outputs"
output_dir.mkdir(exist_ok=True)

m.save(str(output_dir/"lake_victoria_map.html"))

print("Interactive map saved successfully.")
```
